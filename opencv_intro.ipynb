Detekcja prostokątów

```python
phone = cv2.imread('phone.jpg')
phone = imutils.resize(phone, height = 400)
cv2_imshow(phone)
```

Konwersja do skaliszarosci

```python
gray_image = cv2.cvtColor(src = phone, code = cv2.COLOR_BGR2GRAY)
cv2_imshow(gray_image)
```

detekcja krawedzi

```python
edges1 = cv2.Canny(image = gray_image, threshold1 = 70, threshold2 = 200)
cv2_imshow(edges1)
```

```python
gray_image = cv2.GaussianBlur(src = gray_image, ksize = (5, 5), sigmaX = 0)
cv2_imshow(gray_image)
```

```python
edges = cv2.Canny(image = gray_image, threshold1 = 70, threshold2 = 200)
cv2_imshow(edges)
```

detekcja konturu

```python
contours = cv2.findContours(image = edges.copy(),
                            mode = cv2.RETR_TREE,
                            method = cv2.CHAIN_APPROX_SIMPLE)
contours = imutils.grab_contours(contours)
contours = sorted(contours, key = cv2.contourArea, reverse = True)[:10]
```

```python
cnt1 = cv2.drawContours(image = phone.copy(), contours = [contours[0]], contourIdx = -1,
                        color = (0, 0, 255), thickness = 3)
cv2_imshow(cnt1)
```

```python
cnt2 = cv2.drawContours(image = phone.copy(), contours = [contours[2]], contourIdx = -1,
                        color = (0, 0, 255), thickness = 3)
cv2_imshow(cnt2)
```

```python
cnt3 = cv2.drawContours(image = phone.copy(), contours = [contours[3]], contourIdx = -1,
                        color = (0, 0, 255), thickness = 3)
cv2_imshow(cnt3)
```

Detekcja prosotkata

```python
screen_contour = None

for contour in contours:
  perimeter = cv2.arcLength(curve = contour, closed = True)
  approx = cv2.approxPolyDP(curve = contour,epsilon = 0.15 * perimeter, closed = True)
  if len(approx) == 4:
    screen_contour = approx
    break
```
